## **cargar webscraping pandas**

Pandas es una biblioteca de Python utilizada principalmente para **trabajar, organizar, analizar y manipular datos** de una manera sencilla y eficiente. Es muy utilizada en áreas como **análisis de datos, inteligencia artificial, ciencia de datos, estadística** y desarrollo de aplicaciones que necesitan procesar grandes cantidades de información. Pandas permite trabajar con estructuras de datos similares a **tablas de Excel o bases de datos**, siendo sus principales estructuras **Series** y **DataFrame**. Un **DataFrame** puede contener filas y columnas con diferentes tipos de información, como nombres, edades, precios, fechas, categorías, etc.

```python
import pandas as pd
import requests
from io import StringIO

url = "https://es.wikipedia.org/wiki/Anexo:Pa%C3%ADses_y_territorios_dependientes_por_poblaci%C3%B3n"

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 Chrome/139.0 Safari/537.36"
}

respuesta = requests.get(url, headers=headers)

print("Código:", respuesta.status_code)

df_web = pd.read_html(StringIO(respuesta.text))[0]

print(df_web.head())

```
