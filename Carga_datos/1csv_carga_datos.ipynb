Es muy utilizada en áreas como **análisis de datos, inteligencia artificial, ciencia de datos, estadística** y desarrollo de aplicaciones que necesitan procesar grandes cantidades de información. Pandas permite trabajar con estructuras de datos similares a **tablas de Excel o bases de datos**, siendo sus principales estructuras **Series** y **DataFrame**. Un **DataFrame** puede contener filas y columnas con diferentes tipos de información, como nombres, edades, precios, fechas, categorías, etc.

```python
#pip install pandas
#pip install numpy



import pandas as pd
import numpy as np

from sklearn.preprocessing import LabelEncoder, StandardScaler, MinMaxScaler
from imblearn.over_sampling import SMOTE


# Cargar CSV
df = pd.read_csv('dataset_ventas.csv')

print(df.head())
print(df.shape)


# 1. Valores faltantes y duplicados

print("Valores nulos:")
print(df.isnull().sum())

print("Duplicados:")
print(df.duplicated().sum())


# Convertir fecha
df['fecha'] = pd.to_datetime(df['fecha'])


# 2. Codificación de variables categóricas

# LabelEncoder para ciudad
le = LabelEncoder()

df['ciudad'] = le.fit_transform(df['ciudad'])


# One-Hot Encoding para categoria y producto
df = pd.get_dummies(
    df,
    columns=['categoria', 'producto']
)


# Crear variable objetivo
df['venta_alta'] = (
    df['total_venta'] >= df['total_venta'].median()
).astype(int)


# Separar características y variable objetivo

X = df.drop(
    columns=['total_venta', 'venta_alta', 'fecha']
)

y = df['venta_alta']


# 3. Normalización

normalizador = MinMaxScaler()

X[['cantidad', 'precio_unitario']] = normalizador.fit_transform(
    X[['cantidad', 'precio_unitario']]
)


# 4. Balancear con SMOTE

print("Antes de SMOTE:")
print(y.value_counts())

smote = SMOTE(random_state=42)

X_resampled, y_resampled = smote.fit_resample(X, y)

print("Después de SMOTE:")
print(y_resampled.value_counts())


# 5. Dataset final

df_final = X_resampled.copy()

df_final['venta_alta'] = y_resampled

print("\nDataset final:")
print(df_final.head())

print("\nDimensiones finales:")
print(df_final.shape)
#df.head()
#df.tail() #muestra las ultimas filas
#df.sample(n=5) #muestra 5 filas aleatorias
#df.info() #informacion del dataset
#df.describe() #estadisticas del dataset

print(df.head()) #imprime las primeras filas del dataset

```
